```python
keys=visa_df.dtypes.keys().tolist()
values=visa_df.dtypes.values.tolist()
cat,num=[],[]
for i,j in zip(keys,values):        # i is names j is dtypes
    if j=='object':
        cat.append(i)
    else:
        num.append(j)
cat
```
