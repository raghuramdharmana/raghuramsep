```python
# PIE CHART
```

```python

- autopct=%a.bf%%           # automatic percentage
- first % is the formatting
- a: how much min width
- b: after point how many decimals
- %% : % symbol
- %1.1f%%: 40.2%
- %2.2f%%: 423.23
```

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
```

```python
plt.pie(values,
        explode=[0.2 for i in range(6)],
        labels=keys,
        autopct='%1.2f%%',
        radius=0.7,
        startangle=45)


plt.savefig('continent_pie.jpg')
plt.show()
```
